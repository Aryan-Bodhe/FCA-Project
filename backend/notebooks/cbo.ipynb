```python
from algorithms.closebyone import CloseByOne
import pandas as pd
```

```python
df = pd.read_csv("fca_benchmark_20x12.csv")
```

```python
def preprocess(df: pd.DataFrame):
    df = df.rename(columns={"Unnamed: 0": "object_id"})
    df = df.set_index("object_id")
    return df.astype(bool)
```

```python
df_p = preprocess(df)
```

```python
algo = CloseByOne()
```

```python
algo.prepare(df_p)
```

```python
concepts = algo.run()
```
